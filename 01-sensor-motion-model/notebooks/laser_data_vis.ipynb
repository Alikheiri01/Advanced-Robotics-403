```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_style('darkgrid')
```

```python
## Reading laser data for different distances

data5 = pd.read_csv('../Data/laser_range_data_5.csv')['range']
data10 = pd.read_csv('../Data/laser_range_data_10.csv')['range']
data15 = pd.read_csv('../Data/laser_range_data_15.csv')['range']
data20 = pd.read_csv('../Data/laser_range_data_20.csv')['range']
data25 = pd.read_csv('../Data/laser_range_data_25.csv')['range']
data30 = pd.read_csv('../Data/laser_range_data_30.csv')['range']
data37 = pd.read_csv('../Data/laser_range_data_37.csv')['range']

```

```python
## Combine all data into one df

df = pd.DataFrame({
    '5cm': data5,
    '10cm': data10,
    '15cm': data15,
    '20cm': data20,
    '25cm': data25,
    '30cm': data30,
    '37cm': data37
})

df.head(10)
```

```python
df.to_csv('laser_data.csv')
```

```python
## Report mean and variance for each distance

rows = []

for col in df.columns:
    mean = df[col].mean()
    std = df[col].std()

    rows.append({
        'distance': col,
        'mean': mean,
        'std': std
    })

df_describe = pd.DataFrame(rows)
```

```python
## Plot pdf and histogram

def plot_pdf_and_histogram(column_name):

    plt.figure(figsize=(10, 6))
    
    plt.figure(figsize=(8, 5))
    sns.histplot(column_name,bins = 30 ,color='skyblue', kde=False)
    plt.title(f"Histogram of {column_name.name}")
    plt.xlabel("Range Value")
    plt.ylabel("Count")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5))
    sns.kdeplot(column_name, color='blue', linewidth=2, fill=True)
    plt.title(f"PDF of {column_name.name}")
    plt.xlabel("Range Value")
    plt.ylabel("Density")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
for i in df.columns:
    plot_pdf_and_histogram(df[i])
```
