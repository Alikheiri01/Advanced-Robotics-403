```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_style('darkgrid')
```

```python
data2 = pd.read_csv('../Data/velocities_2.5.csv')['velocity (m/s)']
data5 = pd.read_csv('../Data/velocities_5.csv')['velocity (m/s)']
data7 = pd.read_csv('../Data/velocities_7.5.csv')['velocity (m/s)']
```

```python
## Combine all data into one df

df = pd.DataFrame({
    't = 2.5': data2,
    't = 5': data5,
    't = 7.5': data7,
   
})

df.head(10)
```

```python
df.to_csv('velocity_data.csv')
```

```python
## Report mean and variance for 

rows = []

for col in df.columns:
    mean = df[col].mean()
    std = df[col].std()

    rows.append({
        't': col,
        'mean': mean,
        'std': std
    })

df_describe = pd.DataFrame(rows)

df_describe.head(10)
```

```python
## Plot pdf and histogram

def plot_pdf_and_histogram(column_name):

    plt.figure(figsize=(10, 6))
    
    plt.figure(figsize=(8, 5))
    sns.histplot(column_name,bins = 30 ,color='skyblue', kde=False)
    plt.title(f"Histogram of {column_name.name}")
    plt.xlabel("Range Value")
    plt.ylabel("Count")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5))
    sns.kdeplot(column_name, color='blue', linewidth=2, fill=True)
    plt.title(f"PDF of {column_name.name}")
    plt.xlabel("Range Value")
    plt.ylabel("Density")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
for i in df.columns:
    plot_pdf_and_histogram(df[i])
```
